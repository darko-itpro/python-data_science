# Traitement d'images

Ce notebook illustre les capacités de numpy pour le traitement d'images associé à Matplotlib pour la lecture et l'affichage.

**WIP** ce notebook est en cours de réalisation.

```python
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path
```

```python
path_to_image = Path('') # Définissez un chemin vers une image jpg

img = plt.imread(path_to_image)
```

```python
plt.imshow(img)
plt.show()
```

```python
gris = (img @ [0.2126, 0.7152, 0.0722]).astype(np.uint8)
```

```python
plt.imshow(gris, cmap="gray")
```

```python
plt.imshow(img[100:300, 100:300])
```
